```python
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import pandas as pd
```

```python
iris = load_iris()
```

```python
X = iris.data
y = iris.target
```

```python
X.shape
```

```python
#Visualize
sns.scatterplot(x=X[:,0],y=X[:,2], c=y)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
#optional  - dimensionality reducition using PCA
```

```python
#Elbow method
wcss = []
for k in range(1,11):
    kmeans = KMeans(n_clusters = k)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
sns.lineplot(x=range(1,11), y = wcss, marker='o')
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
#Kmeans 
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=42)
labels = kmeans.fit_predict(X_scaled)

sns.scatterplot(x=X[:, 0], y=X[:, 2], c=labels)
sns.scatterplot(x= kmeans.cluster_centers_[:,0], y=kmeans.cluster_centers_[:,1],marker="x",c="red", s=200)
```
